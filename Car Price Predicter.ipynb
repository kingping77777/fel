```python
X=df.drop(columns=["price"] )
Y=df["price"]
```

```python
X
```

```python
df.columns
```

```python
X_one_encode= pd.get_dummies(X, columns=['model','transmission','fuelType'],drop_first=True)
print(X_one_encode)
```

```python
X_one_encode.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder

columns=['model','transmission','fuelType']
Xlable=X.copy()
label_encoders={}
for col in columns:
   le= LabelEncoder()
   Xlable[col]= le.fit_transform(Xlable[col].astype(str))
   label_encoders[col]=le
```

```python
Xlable
```

```python
Xlable["model"].value_counts
```

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
numerical_cols=['year','mileage','tax','mpg', 'engineSize' ]
X_one_encode[numerical_cols]= scaler.fit_transform(X_one_encode[numerical_cols])
```

```python
X_one_encode.astype(int)
```

```python
Xlable.columns
```

```python
Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']]=scaler.fit_transform(Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']])
```

```python
Xlable
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import mean_absolute_error , mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test= train_test_split(X_one_encode, Y, test_size=0.33, random_state=42)
```

```python
model= LinearRegression ()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2= r2_score(y_test,y_pred)
r2
```

```python
import joblib

# Save the trained model
joblib.dump(model, "car_price_model.pkl")

# Save the scaler used for the 5 numerical columns
joblib.dump(scaler, "scaler_model1.pkl")

# Save the exact feature columns
joblib.dump(X_one_encode.columns.tolist(), "feature_columns.pkl")

# Save numerical column names
joblib.dump(numerical_cols, "numerical_cols.pkl")

print("✅ Model 1 files saved successfully!")
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1)/(n-p-1))
print("Adjusted R Square score:",adjusted_r2)
```
